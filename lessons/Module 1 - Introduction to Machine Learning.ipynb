### Matrix multiplication implementation for educational purposes

```python
a = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

b = np.array([
    [7, 8, 9],
    [10, 11, 12]
])
```

```python
def mat_mul(A: np.array, B: np.array) -> np.array:
    assert A.shape[1] == B.shape[0]

    result_mat = np.zeros(shape=(A.shape[0], B.shape[1]))

    for i in range(A.shape[0]):
        for j in range(B.shape[1]):
            result_mat[i, j] = (A[i] * B[:, j]).sum()

    return result_mat
```

```python
mat_mul(a, b)
```
